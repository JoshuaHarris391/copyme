```python
import json

sample_text = """

Ahoy there, ye scurvy dogs! *hic* Let me tell ye about the time I sailed the seven seas with nothin' but a bottle o' whiskey and me lucky shamrock! By the saints, the waves were as tall as me dear old mam's temper, they were! And wouldn't ye know it, the captain - bless his barnacle-covered soul - he says to me, "Paddy," he says, "ye've had enough grog fer one night!" *hic* But I told him straight, I did - "Captain, when an Irishman's had enough, the bottle's empty and the sun's come up twice!" Aye, those were the days, me hearties... now where'd I put me other boot?

*hic* Now, let me tell ye about the great storm of '87! We were three weeks out from port, we were, and the sky turned blacker than a priest's cassock at midnight! The wind was howlin' like a banshee at a wake, and the rain - oh, the rain came down sideways, it did! Me shipmates were prayin' to every saint they could remember, but not old Paddy - no sir! I grabbed hold of the mast with one hand and me whiskey with the other, and I sang "Danny Boy" at the top of me lungs until the storm gave up and went home! Even Neptune himself couldn't dampen an Irish spirit, I tell ye! *hic*

And don't even get me started on the time we found that treasure island! The map was drawn on the back of an old pub receipt - fitting, if ye ask me! We dug fer hours under the blazing sun, and what did we find? A chest full of gold doubloons and - ye won't believe this - a perfectly preserved bottle of aged Irish whiskey! The lads wanted to split the gold, but I told 'em, "Boys, ye can have the shiny bits - this here liquid gold is worth more than all the treasure in the Caribbean!" *hic* They thought I was mad, they did, but who's laughin' now? Not them - they spent their gold years ago, but I've still got the memories of that magnificent whiskey! Aye, 'twas smoother than a leprechaun's promise, it was!

"""

results = analyze_linguistic_style(sample_text)

print("=== Quantitative Metrics ===")
for metric, value in results["quantitative"].items():
    print(f"  {metric}: {value}")

print("\n=== Qualitative Assessments ===")
for assessment, value in results["qualitative"].items():
    print(f"  {assessment}: {value}")

print("\n=== Words Data ===")
for key, value in results["words"].items():
    print(f"  {key}: {value}")
```
